## 常见图形种类及意义
折线图plot

散点图scatter（关系/规律）

柱状图bar（统计/对比）

直方图histogram（数据分布状况）

饼图pie（占比）

### 一、散点图：
应用场景：探究不同变量之间的关系

```python
import matplotlib.pyplot as plt

# 需求探究房屋面积和房屋价格关系
# 1.准备数据
x = [225.98, 240.07, 253.14, 457.85, 241.58, 301.01, 20.67, 288.64, 163.56, 120.66, 207.83, 342.755, 147.9, 53.06,
     224.72, 29.51, 21.61, 483.21, 245.25, 399.25, 343.35]
y = [196.63, 203.88, 210.75, 372.74, 202.41, 247.61, 24.9, 239.34, 140.32, 104.15, 175.84, 288.23, 128.79, 49.64,
     191.74, 33.1, 30.74, 400.02, 205.35, 330.64, 283.45]

# 2.创建画布
plt.figure(figsize=(20, 8), dpi=100)
# 3.绘制图像
plt.scatter(x, y)
# 4.显示图像
plt.show()
```

## 二、柱状图
需求1-对比每部电影的票房收入

```python
# 1.准备数据
movie_names = ['雷神3：诸神黄昏', '正义联盟', '东方快车谋杀案', '寻梦环游记', '全球风暴', '降魔传', '追捕', '七十七天', '密战', '狂兽', '其它']
tickets = [73853, 57767, 22354, 15969, 14839, 8725, 8716, 8318, 7916, 6764, 52222]

# 2.创建画布
plt.figure(figsize=(20, 8), dpi=100)
# 3.绘制图像
x_movie_name = range(len(movie_names))
plt.bar(x_movie_name, tickets, color=['b', 'r', 'g', 'y', 'c', 'm', 'y', 'k', 'c', 'g', 'b'])
# 修改刻度
plt.xticks(x_movie_name, movie_names)
# 添加网格
plt.grid(linestyle="--", alpha=0.5)
plt.title("电影票房收入对比")
# 4.显示图像
plt.show()
```

需求2-如何对比电影票房才更有说服力

```python
# 1.准备数据
movie_name = ['雷神3：诸神黄昏', '正义联盟', '寻梦环游记']
first_day = [10587.6, 10062.5, 1275.7]
first_weekend = [36224.9, 34479.6, 11830]
x = range(len(movie_name))
# 2.创建画布
plt.figure(figsize=(20, 8), dpi=100)
# 3.绘制图像
plt.bar(x, first_day, width=0.2, label="首日票房")
plt.bar([i + 0.2 for i in x], first_weekend, width=0.2, label="首周票房")
# 显示图例
plt.legend()
# 修改刻度
plt.xticks([i + 0.1 for i in x], movie_name)
# 添加网格
plt.grid(linestyle="--", alpha=0.5)
plt.title("电影票房收入对比")
# 4.显示图像
plt.show()
```
