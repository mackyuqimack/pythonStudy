### 3.生成固定范围的数组
#### numpy.linspace

numpy.linspace 函数用于创建一个一维数组，数组是一个等差数列构成的（左闭右闭），格式如下：

```
np.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None)
```

参数说明：

| 参数       | 描述                                                         |
| :--------- | :----------------------------------------------------------- |
| start   | 序列的起始值                                                 |
| stop     | 序列的终止值，如果`endpoint`为`true`，该值包含于数列中       |
| num      | 要生成的等步长的样本数量，默认为`50`                         |
| endpoint | 该值为 `true` 时，数列中包含`stop`值，反之不包含，默认是True。 |
| retstep  | 如果为 True 时，生成的数组中会显示间距，反之不显示。         |
| dtype    | `ndarray` 的数据类型                                         |



````python
np.linspace(0, 10, 5)
````

#### numpy.arange

numpy 包中的使用 arange 函数创建数值范围并返回 ndarray 对象，函数格式如下：

```
numpy.arange(start, stop, step, dtype)
```

根据 start 与 stop 指定的范围以及 step 设定的步长，生成一个 ndarray。

参数说明：

| 参数    | 描述                                                         |
| :------ | :----------------------------------------------------------- |
| start | 起始值，默认为`0`                                            |
| stop  | 终止值（不包含）                                             |
| step  | 步长，默认为`1`                                              |
| dtype | 返回`ndarray`的数据类型，如果没有提供，则会使用输入数据的类型。 |

````python
np.arange(0, 11, 5)
````

### 4.生成随机数

#### 均匀分布
每组可能性相等

##### np.random.rand(d0,d1,....,dn)
返回[0.0，1.1]内一组均匀分布的数

##### np.random.uniform(low=0.0,high=1.0,size=None)
功能:从一个均匀分布[low,high)中随机采样，注意定义域是左闭右开，即包含low，不包含high。

````python
arr1 = np.random.uniform(low=-1,high=1,size=1000000)
````

````python
arr1
````

````python
import matplotlib.pyplot as plt

# 1.创建画布
plt.figure(figsize=(20, 8), dpi=80)
# 2.绘制直方图
# density 显示频率
plt.hist(arr1, bins=1000)
# 3.显示图像
plt.show()
````

#### 正态分布
1.正态分布是一种概率分布。正态分布是具有两个参数μ和a的连续型随机变量的分布，第一参数μ是服从正态分布的随机变量的均值，第二个参数σ是此随机变量的方差，所以正态分布记作N(μ, σ )(miu,sigema)。


2.正态分布的应用：生活、生产与科学实验中很多随机变量的概率分布都可以近似地用正态分布来描述。

3.正态分布的特点：**μ决定了其位置，其标准差σ决定了分布的幅度**。当μ= 0,σ = 1时的正态分布是标准正态分布。

##### 方差
在概率论和统计学中衡量一组数据离散程度的度量


##### 标准差与方差的意义
可以理解成数据的一个离散程度的衡量

- np.random.randn(d0,d1,....,dn)

功能：从标准正态分布中返回一个或多个样本值

- np.random.normal(loc=0.0,scale=1.0,size=None)

loc:float

此概率分布的均值（对应着整个分布的中心centre）

scale:float

此概率分布的标准差（对应于分布的宽度，scale越大越矮胖，scale越小越瘦高）

size:int or tuple of ints

输出的shape，默认为None，只输出一个值

- np.random.standard_normal(size=None)

返回指定形状的标准正态分布的数组。

````python
arr2 = np.random.normal(loc=1.75, scale=0.1, size=1000000)
````

````python
arr2
````

````python
# 1.创建画布
plt.figure(figsize=(20, 8), dpi=80)
# 2.绘制直方图
# density 显示频率
plt.hist(arr2, bins=1000)
# 3.显示图像
plt.show()
````

### 6.逻辑运算

````python
stock_change = np.random.normal(loc=0, scale=1, size=(8,10))
````

````python
stock_change
````

````python
# 逻辑判断，如果大于0.5标记True,否则标记False
stock_change > 0.5
````

````python
# 布尔索引
# 我们可以通过一个布尔数组来索引目标数组。
stock_change[stock_change > 0.5]
````

````python
# 将大于0.5的值赋值为1.1
stock_change[stock_change > 0.5] = 1.1
````

````python
stock_change
````

### 7.通用判断函数
- np.all()
 
- np.any()

````python
# 判断stock_change[0:2, 0:5]是否全是大于0
stock_change[0:2, 0:5] > 0
````

````python
np.all(stock_change[0:2, 0:5] > 0)
````

````python
# 判断前5个值这段期间是否有大于0
np.any(stock_change[:5, :] > 0)
````
